# **1. Data Cleansing**

**1. Mistake During Data Entry**

***Deteksi Mistake During Data Entry 1.1 pada Bulan_Evaluasi***

```python
data.groupby('Bulan_Evaluasi').size()
```

Proses 1.1

```python
data["Bulan_Evaluasi"] = data["Bulan_Evaluasi"].replace(["januari"], "Januari")
data["Bulan_Evaluasi"] = data["Bulan_Evaluasi"].replace(["februari"], "Februari")
data["Bulan_Evaluasi"] = data["Bulan_Evaluasi"].replace(["Mart", "MARET", "maret"], "Maret")
data["Bulan_Evaluasi"] = data["Bulan_Evaluasi"].replace(["april"], "April")
data["Bulan_Evaluasi"] = data["Bulan_Evaluasi"].replace(["mei", "MEI"], "Mei")
data["Bulan_Evaluasi"] = data["Bulan_Evaluasi"].replace(["juni"], "Juni")
data["Bulan_Evaluasi"] = data["Bulan_Evaluasi"].replace(["Jul", "JULI", "juli"], "Juli")
data["Bulan_Evaluasi"] = data["Bulan_Evaluasi"].replace(["Agutus", "agustus"], "Agustus")
data["Bulan_Evaluasi"] = data["Bulan_Evaluasi"].replace(["september"], "September")
data["Bulan_Evaluasi"] = data["Bulan_Evaluasi"].replace(["Okotber", "oktober"], "Oktober")
data["Bulan_Evaluasi"] = data["Bulan_Evaluasi"].replace(["desember"], "Desember")
```

***Deteksi Mistake During Data Entry 1.3 Pada Pendidikan***

# **2. DATA TRANSFORMATION**

**1. Transformation Input Data**

```python
data_clean = data.copy()
data_clean.head(10)
```

**2. Scaling**

2.1 Min-Max Scaling (Manual)

```python
data = data_clean.copy()
data['Gaji_Juta'] = (data['Gaji_Juta'] - data['Gaji_Juta'].min()) / (data['Gaji_Juta'].max() - data['Gaji_Juta'].min())
data['Gaji_Juta'] = data['Gaji_Juta'].round(3)
data.head(10)
```

2.2 Min-Max Scaling (Library)

```python
from sklearn.preprocessing import MinMaxScaler
data[['Gaji_Juta']] = MinMaxScaler().fit_transform(data[['Gaji_Juta']])
data.head(10)
```

2.3 Z-score Standardization (Manual)

```python
data = data_clean.copy()
data['Gaji_Juta'] = (data['Gaji_Juta'] - data['Gaji_Juta'].mean()) / data['Gaji_Juta'].std()
data['Gaji_Juta'] = data['Gaji_Juta'].round(3)
data.head(10)
```

2.4 Z-score Standardization (Library)

```python
from sklearn.preprocessing import StandardScaler
data[['Gaji_Juta']] = StandardScaler().fit_transform(data[['Gaji_Juta']])
data.head(10)
```

2.5 Robust Scaling (Manual)

```python
data = data_clean.copy()
q1 = data['Gaji_Juta'].quantile(0.25)
q3 = data['Gaji_Juta'].quantile(0.75)
iqr = q3 - q1
data['Gaji_Juta'] = (data['Gaji_Juta'] - data['Gaji_Juta'].median()) / iqr
data.head(10)
```

2.6 Robust Scaling (Library)

```python
from sklearn.preprocessing import RobustScaler
data[['Gaji_Juta']] = RobustScaler().fit_transform(data[['Gaji_Juta']])
data.head(10)
```

**3. Encoding**

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
pendidikan = ['SMA', 'D3', 'S1', 'S2']
data = pd.DataFrame({'Pendidikan': pendidikan})
data
```

```python
label_encoder = LabelEncoder()
data['Label_Pendidikan'] = label_encoder.fit_transform(data['Pendidikan'])
pendidikan_map = {'SMA': 0, 'D3': 1, 'S1': 2, 'S2': 3}
data['Label_Pendidikan'] = data['Pendidikan'].map(pendidikan_map)
data
```

**4. Data Discretization**

4.1 Equal-Width

```python
data = data_clean.copy()
data['Skor_Kinerja_Bin'] = pd.cut(data['Skor_Kinerja'], bins=3, labels=['Rendah', 'Sedang', 'Tinggi'])
print(data['Skor_Kinerja_Bin'].value_counts())
data.head(10)
```

4.2 Equal-Frequency

```python
data['Skor_Kinerja_Quantile'] = pd.qcut(data['Skor_Kinerja'], q=3, labels=['Rendah', 'Sedang', 'Tinggi'])
print(data['Skor_Kinerja_Quantile'].value_counts())
data.head(10)
```

**5. Data Measurement and Aggregation**

5.1 Different Unit of Measurement

```python
data = data_clean.copy()
data['Gaji_Juta'] = data['Gaji_Juta']*10**6
data.head(10)
```

5.2 Different Level of Aggregation

```python
data_num = data[['Bulan_Evaluasi', 'Usia', 'Masa_Kerja_Tahun', 'Gaji_Juta', 'Skor_Kinerja']]
data_agg = data_num.groupby('Bulan_Evaluasi').mean()
data_agg
```

5.3 Attribute Construction

```python
data['Usia_Mulai_Kerja'] = data['Usia'] - data['Masa_Kerja_Tahun']
data.head(10)
```

5.4 Data Smoothing

```python
data_smooth = data.copy()
data['Skor_Kinerja_MA'] = data['Skor_Kinerja'].rolling(window=3).mean()
data['Gaji_Juta_ES'] = data['Gaji_Juta'].ewm(alpha=0.3).mean()
data.head(10)
```
